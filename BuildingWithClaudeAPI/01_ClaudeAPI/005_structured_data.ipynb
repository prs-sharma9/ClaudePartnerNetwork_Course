````python
# importing claude helper
from claude_helper import add_assistant_message, add_user_message, chat
````

````python
# This will generate the json code but there will be comments and explaination in the output produced by claude
# we want to generate the required structed data with any comments or markdown.

messages = []

add_user_message(messages, "Generate a very short event bridger rule as json")

chat(messages)
````

````python
# To generate output without any markdown we can use : Prefilled Assistant message and Stop Sequences.

messages = []

add_user_message(messages, "Generate a very short event bridger rule as json")

add_assistant_message(messages, "```json")

text = chat(messages, stop_sequences=["```"])

text
````
