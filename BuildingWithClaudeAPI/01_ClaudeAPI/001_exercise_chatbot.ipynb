```python
from claude_helper import chat, add_assistant_message, add_user_message
```

```python
# Write logic to take input from user

def start_chat(messages):
    print("Start chatting with Claude. (Type: 'exit' to end chat)")
    while(True):

        # Take user input
        user_prompt = input("You: ")
        if(user_prompt == 'exit'):
            print("Thank you!\n")
            break
        add_user_message(messages, user_prompt)
        print(f"You: {user_prompt}\n")

        # send prompt to model
        ai_response = chat(messages)

        # add response to chat history
        add_assistant_message(messages, ai_response)

        print(f"Assistant: {ai_response}\n")
```

```python
start_chat([])
```
