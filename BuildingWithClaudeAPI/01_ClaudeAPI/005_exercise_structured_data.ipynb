````python
from claude_helper import chat, add_user_message, add_assistant_message
````

````python
messages = []

prompt = """
Generate three different sample AWS CLI commands. Each should be very short
"""

assistant_prefilling = "Here are the three commands in a single block without any comments:\n```bash"

add_user_message(messages, prompt)

add_assistant_message(messages, assistant_prefilling)

text = chat(messages, stop_sequences=["```"])

text.strip()
````
